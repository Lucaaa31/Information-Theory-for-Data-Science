# Exercise 2 – How unpredictable is a sporting event?

## Question 1: Select three sporting events: two with the same number of outcomes and one with a different number.

The sporting events selected with the same number of outcomes are:
- Barcelona - Inter Milan, Champions League Semifinal
    - Odds: Home Win $1.44$, Draw $4.50$, Away Win $7.50$
- Inter Milan - AC Milan, Serie A
    - Odds: Home Win $2.15$, Draw $3.40$, Away Win $3.50$

The match with differents outcomes is:
- Jannik Sinner - Carlos Alcaraz, Wimbledon Final
    - Odds: Player 1 Win $1.91$, Player 2 Win $1.91$

## Question 2: From decimal odds $o_i$, form $q_i = \frac{1}{o_i}$ and normalize $p_i = \frac{q_i}{\sum_j q_j}$.

#### 1. Barcelona vs. Inter Milan (Champions League Semifinal)

**Unnormalized Probabilities ($q_i$)**
* $q_1 = \frac{1}{1.44} \approx 0.69444$
* $q_2 = \frac{1}{4.50} \approx 0.22222$
* $q_3 = \frac{1}{7.50} \approx 0.13333$

**Normalized Probabilities ($p_i$)**
* $p_{\text{Home}} = \frac{0.69444}{1.05000} \approx 0.66138 \quad \mathbf{(66.14\%)}$
* $p_{\text{Draw}} = \frac{0.22222}{1.05000} \approx 0.21164 \quad \mathbf{(21.16\%)}$
* $p_{\text{Away}} = \frac{0.13333}{1.05000} \approx 0.12698 \quad \mathbf{(12.70\%)}$

#### 2. Inter Milan vs. AC Milan (Serie A)

**Unnormalized Probabilities ($q_i$)**
* $q_1 = \frac{1}{2.15} \approx 0.46512$
* $q_2 = \frac{1}{3.40} \approx 0.29412$
* $q_3 = \frac{1}{3.50} \approx 0.28571$


**Normalized Probabilities ($p_i$)**
* $p_{\text{Home}} = \frac{0.46512}{1.04495} \approx 0.44511 \quad \mathbf{(44.51\%)}$
* $p_{\text{Draw}} = \frac{0.29412}{1.04495} \approx 0.28147 \quad \mathbf{(28.15\%)}$
* $p_{\text{Away}} = \frac{0.28571}{1.04495} \approx 0.27342 \quad \mathbf{(27.34\%)}$

#### 3. Jannik Sinner vs. Carlos Alcaraz (Wimbledon Final)

**Unnormalized Probabilities ($q_i$)**
* $q_1 = \frac{1}{1.91} \approx 0.52356$
* $q_2 = \frac{1}{1.91} \approx 0.52356$


**Normalized Probabilities ($p_i$)**
* $p_{\text{Player 1}} = \frac{0.52356}{1.04712} = 0.50000 \quad \mathbf{(50.00\%)}$
* $p_{\text{Player 2}} = \frac{0.52356}{1.04712} = 0.50000 \quad \mathbf{(50.00\%)}$


## Question 3: Plot the three pmfs and compute $H(X)$.

`````
% 1. Define Odds
odds_barca_inter    = [1.44, 4.50, 7.50];
odds_inter_milan    = [2.15, 3.40, 3.50];
odds_sinner_alcaraz  = [1.91, 1.91];

% 2. Compute Normalized Probabilities p_i
p1 = (1 ./ odds_barca_inter)    / sum(1 ./ odds_barca_inter);
p2 = (1 ./ odds_inter_milan)    / sum(1 ./ odds_inter_milan);
p3 = (1 ./ odds_sinner_alcaraz)  / sum(1 ./ odds_sinner_alcaraz);

% Shannon Entropy Function (in bits)
shannon_entropy = @(p) -sum(p .* log2(p));

% Compute Entropy H(X)
H1 = shannon_entropy(p1);
H2 = shannon_entropy(p2);
H3 = shannon_entropy(p3);

% 4. Compute Normalized Entropy: H_norm = H(X) / log2(M)
% M represents the number of possible outcomes for each event
H1_norm = H1 / log2(length(p1)); % M = 3
H2_norm = H2 / log2(length(p2)); % M = 3
H3_norm = H3 / log2(length(p3)); % M = 2

% Display Numerical Results in Console
fprintf('--- ENTROPY RESULTS ---\n');
fprintf('Barcelona vs Inter  (M=3): H = %.4f bits | H_norm = %.4f\n', H1, H1_norm);
fprintf('Inter vs AC Milan   (M=3): H = %.4f bits | H_norm = %.4f\n', H2, H2_norm);
fprintf('Sinner vs Alcaraz   (M=2): H = %.4f bits | H_norm = %.4f\n\n', H3, H3_norm);

% 3. Standard MATLAB Plotting
figure;

events = { ...
    'Barcelona vs Inter', p1, {'Home', 'Draw', 'Away'}, H1_norm; ...
    'Inter vs AC Milan',  p2, {'Home', 'Draw', 'Away'}, H2_norm; ...
    'Sinner vs Alcaraz',  p3, {'Player 1', 'Player 2'}, H3_norm ...
};

for idx = 1:3
    subplot(1, 3, idx);
    
    title_str = events{idx, 1};
    p         = events{idx, 2};
    labels    = events{idx, 3};
    hnorm     = events{idx, 4};
    
    % Standard Bar Plot
    bars = bar(categorical(labels, labels, 'Ordinal', true), p);
    
    % Subplot Title with H_norm value
    title(sprintf('%s\n(H_{norm} = %.4f)', title_str, hnorm));
    
    if idx == 1
        ylabel('Probability p_i');
    end
    ylim([0, 0.8]);
    
    % Add values on top of bars
    xtips = bars.XEndPoints;
    ytips = bars.YEndPoints;
    text(xtips, ytips, string(num2str(p', '%.4f')), ...
         'HorizontalAlignment', 'center', ...
         'VerticalAlignment', 'bottom');
end
`````
